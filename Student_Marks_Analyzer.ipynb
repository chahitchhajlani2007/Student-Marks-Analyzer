```python
import numpy as np

subjects = np.array(["Maths", "Physics", "Chemistry", "English", "Computer"])

marks = np.random.randint(35, 100, (50, 5))

print("Subjects:", subjects)
print("Student Marks:\n", marks)
```

```python
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total Elements:", marks.size)
print("Data Type:", marks.dtype)

print("Minimum Marks:", marks.min())
print("Maximum Marks:", marks.max())
print("Overall Average:", marks.mean())
```

```python
print("First 10 Students:\n", marks[:10])
print("Last Subject Marks:\n", marks[:, -1])
print("First 3 Subjects:\n", marks[:, :3])
print("Students 11 to 20:\n", marks[10:20])
```

```python
below_40 = marks[marks < 40]
above_90 = marks[marks > 90]

print("Marks Below 40:", below_40)
print("Marks Above 90:", above_90)

# Grade bands
A = marks[marks >= 90]
B = marks[(marks >= 75) & (marks < 90)]
C = marks[(marks >= 60) & (marks < 75)]
D = marks[(marks >= 40) & (marks < 60)]
F = marks[marks < 40]

print("Grade A:", A.size)
print("Grade B:", B.size)
print("Grade C:", C.size)
print("Grade D:", D.size)
print("Grade F:", F.size)
```

```python
for i in range(len(subjects)):
    topper = marks[:, i].max()
    print(subjects[i], "Topper Marks:", topper)

    pass_count = np.sum(marks[:, i] >= 40)
    pass_percentage = (pass_count / 50) * 100

    print("Pass Percentage:", pass_percentage, "%")
```

```python
total_marks = marks.sum(axis=1)
average_marks = marks.mean(axis=1)

topper_index = np.argmax(total_marks)

print("Overall Topper: Student", topper_index + 1)
print("Total Marks:", total_marks[topper_index])
print("Average Marks:", average_marks[topper_index])

failed_students = np.any(marks < 40, axis=1)

print("Failed Student Numbers:",
      np.where(failed_students)[0] + 1)

print("Total Failed Students:", np.sum(failed_students))
```

```python
print("STUDENT MARKS ANALYZER")
print("----------------------")
print("Total Students:", marks.shape[0])
print("Total Subjects:", marks.shape[1])
print("Overall Average:", round(marks.mean(), 2))
print("Highest Mark:", marks.max())
print("Lowest Mark:", marks.min())
print("Students Passed All Subjects:", np.sum(~failed_students))
print("Students Failed in At Least One Subject:", np.sum(failed_students))
```
